```python
import matplotlib.pyplot as plt
import numpy as np
import sqlite3 
import pandas as pd
```

```python
conn = sqlite3.connect("galaxy.db")
cursor = conn.cursor()
```

```python
cursor.execute("""CREATE TABLE IF NOT EXISTS galaxy(
ID INTEGER PRIMARY KEY,
target TEXT,
exposure REAL,
observer TEXT,
filter_name TEXT
)""")
conn.commit()
```

```python
class galaxy :
    def __init__(self , target, exposure, observer,filter_name ) :
        self.target = target
        self.exposure = exposure
        self.observer = observer
        self.filter_name = filter_name
    def save_to_db(self, cursor, conn):
        cursor.execute("""
        INSERT INTO galaxy
        (target, exposure, observer, filter_name)
        VALUES(?,?,?,?)""",
                      (self.target,
                       self.exposure,
                      self.observer,
                      self.filter_name))
        conn.commit()
        
```

```python
def add_observation(cursor , conn):
    target = input("Target: ")
    try: 
        exposure = float(input("Exposure time(Second)"))
    except ValueError:
        print("Exposure time should be a int or float")
        return
    observer = input("Name of observer")
    filter_name = input("Filter name: ")
    g = galaxy(target ,exposure ,observer ,filter_name )
    g.save_to_db(cursor , conn)
    
def show_observation(conn) :
    cursor.execute("""SELECT * FROM galaxy""")
    rows = cursor.fetchall()
    for i in rows :
        print(i)
        
def pandas_observation(conn):
    df = pd.read_sql_query("SELECT * FROM galaxy", conn)
    print(df)
    
def statistic_exposure(conn):
    df = pd.read_sql_query("SELECT * FROM galaxy", conn)

    print("Total exposure :", df['exposure'].sum())
    print("Average exposure :", df['exposure'].mean())
    print("maximum exposure :", df['exposure'].max())
    print("Minimum exposure :", df['exposure'].min())
    
def search_observation(conn):
    df = pd.read_sql_query("SELECT * FROM galaxy", conn)
    user = input("Target: ")
    found = False
    result = df[df['target'].str.lower() == user.lower()]
    if not result.empty :
        print("Observation found\n")
        print(result)
        found = True
    if not found:
        print("No target name found")
        
def plot_exposure(conn):
    df = pd.read_sql_query("SELECT * FROM galaxy", conn)
    fig, ax = plt.subplots()
    x = df['target']
    y = df['exposure']
    ax.bar(x , y)
    ax.set_xlabel('target')
    ax.set_ylabel('exposure')
    ax.set_title('exposure time by target')
    plt.show()   

def save_to_file(conn):
    df = pd.read_sql_query("SELECT * FROM galaxy", conn)
    filename = input("Your file name: ")
    df.to_csv(f"{filename}.csv", index= False) 
    print("file saved ")
```

```python
while True:
    print("=" * 40)
    print()
    print("Welcome to Observer Log".center(40))
    print()
    print("=" * 40)
    print("1. Add Observation")
    print("2. Show Observation")
    print("3. pandas exposure")
    print("4. statistic exposure")
    print("5. search observation")
    print("6. plot exposure")
    print("7. Save to file")
    print("8. Exit")
    choice = input("Choose:")
    if choice == "1":
        add_observation(cursor , conn)
    elif choice == "2":
        show_observation(conn)
    elif choice == "3":
        pandas_observation( conn)
    elif choice == "4":
        statistic_exposure(conn)
    elif choice == "5":
        search_observation( conn)
    elif choice == "6":
        plot_exposure( conn)
    elif choice == "7":
        save_to_file( conn)
    elif choice == "8":    
        break
    else: 
        print("Please select 1 to 8")
```
